```python
# ==========================================================
# SAFE DIVISION FUNCTION
# ==========================================================

def safe_divide(numerator, denominator):
    """
    Safely divide two pandas Series.

    Returns 0 whenever the denominator is
    0 or missing.
    """

    numerator = pd.to_numeric(numerator, errors="coerce")
    denominator = pd.to_numeric(denominator, errors="coerce")

    result = np.where(
        denominator > 0,
        numerator / denominator,
        0
    )

    return result
```

```python
# ==========================================================
# PERFORMANCE FEATURES
# ==========================================================

def create_performance_features(df):

    df = df.copy()

    # Total Premium
    df["Total_Premium_Lakh"] = (
        df["Farmers_Premium_Lakh"]
        + df["State_Premium_Lakh"]
        + df["GOI_Premium_Lakh"]
    )

    # Premium Per Farmer
    df["Premium_Per_Farmer"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Farmers"]
    )

    # Sum Insured Per Farmer
    df["Sum_Insured_Per_Farmer"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Farmers"]
    )

    # Benefit Per Farmer
    df["Benefit_Per_Farmer"] = safe_divide(
        df["Total_Farmer_Benefit_Lakh"],
        df["Farmers"]
    )

    # Claim Per Farmer
    df["Claim_Per_Farmer"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers"]
    )

    # Government Premium Ratio
    df["Government_Premium_Ratio"] = safe_divide(
        (
            df["State_Premium_Lakh"]
            + df["GOI_Premium_Lakh"]
        ),
        df["Total_Premium_Lakh"]
    )

    # Farmer Premium Ratio
    df["Farmer_Premium_Ratio"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # Claim Efficiency
    df["Claim_Efficiency"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers_Premium_Lakh"]
    )

    return df
```

```python
# ==========================================================
# AIIS MODULE 1
# Insurance Performance Indicators
# ==========================================================

def create_performance_indicators(df):

    """
    Create Insurance Performance Indicators
    """

    df = df.copy()

    # -----------------------------
    # Total Premium
    # -----------------------------
    df["Total_Premium_Lakh"] = (
        df["Farmers_Premium_Lakh"] +
        df["State_Premium_Lakh"] +
        df["GOI_Premium_Lakh"]
    )

    # -----------------------------
    # Premium Per Farmer
    # -----------------------------
    df["Premium_Per_Farmer"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Benefit Per Farmer
    # -----------------------------
    df["Benefit_Per_Farmer"] = safe_divide(
        df["Total_Farmer_Benefit_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Claim Per Farmer
    # -----------------------------
    df["Claim_Per_Farmer"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Sum Insured Per Farmer
    # -----------------------------
    df["Sum_Insured_Per_Farmer"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Government Support Ratio
    # -----------------------------
    df["Government_Support_Ratio"] = safe_divide(
        df["State_Premium_Lakh"] +
        df["GOI_Premium_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Farmer Contribution Ratio
    # -----------------------------
    df["Farmer_Contribution_Ratio"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Claim Efficiency
    # -----------------------------
    df["Claim_Efficiency"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers_Premium_Lakh"]
    )

    # -----------------------------
    # Benefit Efficiency
    # -----------------------------
    df["Benefit_Efficiency"] = safe_divide(
        df["Total_Farmer_Benefit_Lakh"],
        df["Farmers_Premium_Lakh"]
    )

    # -----------------------------
    # Premium Utilization Ratio
    # -----------------------------
    df["Premium_Utilization_Ratio"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Insurance Leverage Ratio
    # -----------------------------
    df["Insurance_Leverage_Ratio"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Total_Premium_Lakh"]
    )

    return df
```

```python
# ==========================================================
# AIIS MODULE 2
# Participation Indicators
# ==========================================================

def create_participation_indicators(df):

    df = df.copy()

    # -----------------------------
    # Loanee Participation Rate
    # -----------------------------
    df["Loanee_Participation_Rate"] = safe_divide(
        df["Application_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Non-Loanee Participation Rate
    # -----------------------------
    df["Non_Loanee_Participation_Rate"] = safe_divide(
        df["Application_Non_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Total Application Ratio
    # -----------------------------
    df["Total_Application_Ratio"] = safe_divide(
        df["Application_Loanee"] +
        df["Application_Non_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Coverage Density
    # -----------------------------
    df["Coverage_Density"] = safe_divide(
        df["Farmers"],
        df["Area_Insured_Thousand_Hectare"]
    )

    # -----------------------------
    # Insurance Unit Density
    # -----------------------------
    df["Insurance_Unit_Density"] = safe_divide(
        df["Insurance_Units"],
        df["Area_Insured_Thousand_Hectare"]
    )

    # -----------------------------
    # Average Area Per Farmer
    # -----------------------------
    df["Average_Area_Per_Farmer"] = safe_divide(
        df["Area_Insured_Thousand_Hectare"],
        df["Farmers"]
    )

    # -----------------------------
    # Farmers Per Insurance Unit
    # -----------------------------
    df["Farmers_Per_Insurance_Unit"] = safe_divide(
        df["Farmers"],
        df["Insurance_Units"]
    )

    # -----------------------------
    # Sum Insured Per Hectare
    # -----------------------------
    df["Sum_Insured_Per_Hectare"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Area_Insured_Thousand_Hectare"]
    )

    return df
```

```python
# ==========================================================
# AIIS - Statistical Summary Report
# ==========================================================

from scipy.stats import skew, kurtosis

def descriptive_summary(df, columns):

    summary = pd.DataFrame(index=columns)

    summary["Mean"] = df[columns].mean()
    summary["Median"] = df[columns].median()
    summary["Std"] = df[columns].std()
    summary["Variance"] = df[columns].var()
    summary["Minimum"] = df[columns].min()
    summary["Maximum"] = df[columns].max()
    summary["Range"] = (
        summary["Maximum"] -
        summary["Minimum"]
    )

    summary["Skewness"] = df[columns].apply(
        lambda x: skew(x.dropna())
    )

    summary["Kurtosis"] = df[columns].apply(
        lambda x: kurtosis(x.dropna())
    )

    summary["Coefficient_of_Variation"] = (
        summary["Std"] /
        summary["Mean"]
    )

    return summary.round(4)
```

```python
summary = descriptive_summary(df, indicator_cols)

display(summary)
```

```python
# ==========================================================
# Distribution Classification
# ==========================================================

def classify_distribution(sk):

    if abs(sk) < 0.5:
        return "Approximately Normal"

    elif abs(sk) < 1:
        return "Moderately Skewed"

    else:
        return "Highly Skewed"

summary["Distribution"] = (
    summary["Skewness"]
    .apply(classify_distribution)
)

display(summary)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

def safe_divide(numerator, denominator):
    """
    Safely divide two pandas Series.

    Returns 0 whenever the denominator is
    0 or missing.
    """

    numerator = pd.to_numeric(numerator, errors="coerce")
    denominator = pd.to_numeric(denominator, errors="coerce")

    result = np.where(
        denominator > 0,
        numerator / denominator,
        0
    )

    return result

def create_performance_indicators(df):

    """
    Create Insurance Performance Indicators
    """

    df = df.copy()

    # Ensure relevant columns are numeric before calculations
    df["Farmers_Premium_Lakh"] = pd.to_numeric(df["Farmers_Premium_Lakh"], errors="coerce")
    df["State_Premium_Lakh"] = pd.to_numeric(df["State_Premium_Lakh"], errors="coerce")
    df["GOI_Premium_Lakh"] = pd.to_numeric(df["GOI_Premium_Lakh"], errors="coerce")
    df["Farmers"] = pd.to_numeric(df["Farmers"], errors="coerce")
    df["Total_Farmer_Benefit_Lakh"] = pd.to_numeric(df["Total_Farmer_Benefit_Lakh"], errors="coerce")
    df["Total_Claim_Paid_Lakh"] = pd.to_numeric(df["Total_Claim_Paid_Lakh"], errors="coerce")
    df["Sum_Insured_Lakh"] = pd.to_numeric(df["Sum_Insured_Lakh"], errors="coerce")

    # -----------------------------
    # Total Premium
    # -----------------------------
    df["Total_Premium_Lakh"] = (
        df["Farmers_Premium_Lakh"] +
        df["State_Premium_Lakh"] +
        df["GOI_Premium_Lakh"]
    )

    # -----------------------------
    # Premium Per Farmer
    # -----------------------------
    df["Premium_Per_Farmer"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Benefit Per Farmer
    # -----------------------------
    df["Benefit_Per_Farmer"] = safe_divide(
        df["Total_Farmer_Benefit_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Claim Per Farmer
    # -----------------------------
    df["Claim_Per_Farmer"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Sum Insured Per Farmer
    # -----------------------------
    df["Sum_Insured_Per_Farmer"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Farmers"]
    )

    # -----------------------------
    # Government Support Ratio
    # -----------------------------
    df["Government_Support_Ratio"] = safe_divide(
        df["State_Premium_Lakh"] +
        df["GOI_Premium_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Farmer Contribution Ratio
    # -----------------------------
    df["Farmer_Contribution_Ratio"] = safe_divide(
        df["Farmers_Premium_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Claim Efficiency
    # -----------------------------
    df["Claim_Efficiency"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Farmers_Premium_Lakh"]
    )

    # -----------------------------
    # Benefit Efficiency
    # -----------------------------
    df["Benefit_Efficiency"] = safe_divide(
        df["Total_Farmer_Benefit_Lakh"],
        df["Farmers_Premium_Lakh"]
    )

    # -----------------------------
    # Premium Utilization Ratio
    # -----------------------------
    df["Premium_Utilization_Ratio"] = safe_divide(
        df["Total_Claim_Paid_Lakh"],
        df["Total_Premium_Lakh"]
    )

    # -----------------------------
    # Insurance Leverage Ratio
    # -----------------------------
    df["Insurance_Leverage_Ratio"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Total_Premium_Lakh"]
    )

    return df

def create_participation_indicators(df):

    df = df.copy()

    # Ensure relevant columns are numeric before calculations
    df["Application_Loanee"] = pd.to_numeric(df["Application_Loanee"], errors="coerce")
    df["Application_Non_Loanee"] = pd.to_numeric(df["Application_Non_Loanee"], errors="coerce")
    df["Farmers"] = pd.to_numeric(df["Farmers"], errors="coerce")
    df["Area_Insured_Thousand_Hectare"] = pd.to_numeric(df["Area_Insured_Thousand_Hectare"], errors="coerce")
    df["Insurance_Units"] = pd.to_numeric(df["Insurance_Units"], errors="coerce")
    df["Sum_Insured_Lakh"] = pd.to_numeric(df["Sum_Insured_Lakh"], errors="coerce")

    # -----------------------------
    # Loanee Participation Rate
    # -----------------------------
    df["Loanee_Participation_Rate"] = safe_divide(
        df["Application_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Non-Loanee Participation Rate
    # -----------------------------
    df["Non_Loanee_Participation_Rate"] = safe_divide(
        df["Application_Non_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Total Application Ratio
    # -----------------------------
    df["Total_Application_Ratio"] = safe_divide(
        df["Application_Loanee"] +
        df["Application_Non_Loanee"],
        df["Farmers"]
    )

    # -----------------------------
    # Coverage Density
    # -----------------------------
    df["Coverage_Density"] = safe_divide(
        df["Farmers"],
        df["Area_Insured_Thousand_Hectare"]
    )

    # -----------------------------
    # Insurance Unit Density
    # -----------------------------
    df["Insurance_Unit_Density"] = safe_divide(
        df["Insurance_Units"],
        df["Area_Insured_Thousand_Hectare"]
    )

    # -----------------------------
    # Average Area Per Farmer
    # -----------------------------
    df["Average_Area_Per_Farmer"] = safe_divide(
        df["Area_Insured_Thousand_Hectare"],
        df["Farmers"]
    )

    # -----------------------------
    # Farmers Per Insurance Unit
    # -----------------------------
    df["Farmers_Per_Insurance_Unit"] = safe_divide(
        df["Farmers"],
        df["Insurance_Units"]
    )

    # -----------------------------
    # Sum Insured Per Hectare
    # -----------------------------
    df["Sum_Insured_Per_Hectare"] = safe_divide(
        df["Sum_Insured_Lakh"],
        df["Area_Insured_Thousand_Hectare"]
    )

    return df

df = create_performance_indicators(df)
df = create_participation_indicators(df)

X = df[indicator_cols].copy()

# Remove rows containing NaN
X = X.dropna()

# Standardize
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns

)
```
